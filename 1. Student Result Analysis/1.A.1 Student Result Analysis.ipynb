# 🎓 Student Result Analysis Project

## 📋 Project Overview

This project analyzes student results from four subjects:

- Math
- Science
- English
- History

The dataset contains 20 students. For each student, the notebook calculates:

- Total marks
- Average marks
- Grade
- Pass/Fail status

### Grading Scale

| Average Score | Grade |
|---:|:---:|
| 75 - 100 | A |
| 60 - 74.99 | B |
| 50 - 59.99 | C |
| 40 - 49.99 | D |
| Below 40 | Fail |

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

data = {
    "Student_ID": range(1001, 1021),
    "Name": [f"Student_{i}" for i in range(1, 21)],
    "Math": np.random.randint(35, 100, 20),
    "Science": np.random.randint(40, 100, 20),
    "English": np.random.randint(30, 98, 20),
    "History": np.random.randint(45, 100, 20)
}

df = pd.DataFrame(data)

df["Total"] = df[["Math", "Science", "English", "History"]].sum(axis=1)
df["Average"] = df["Total"] / 4

df["Grade"] = pd.cut(
    df["Average"],
    bins=[0, 40, 50, 60, 75, 100],
    labels=["Fail", "D", "C", "B", "A"],
    include_lowest=True
)

df.head()
```

## 📊 Summary Statistics

This section shows the overall performance of the class.

It includes:

- Class pass rate
- Number of failed students
- Highest average score
- Lowest average score
- Subject-wise statistics

```python
summary = df[["Math", "Science", "English", "History", "Average"]].describe().round(2)

pass_count = (df["Grade"] != "Fail").sum()
fail_count = (df["Grade"] == "Fail").sum()

print(f"📈 Class Pass Rate: {pass_count}/{len(df)} ({pass_count / len(df) * 100:.1f}%)")
print(f"❌ Fail Count: {fail_count}")
print(f"🏆 Highest Average: {df['Average'].max():.2f}")
print(f"⚠️ Lowest Average: {df['Average'].min():.2f}")

summary
```

## 📈 Visualizations

The following charts are used to understand student performance:

1. **Grade Distribution**  
   Shows how many students received A, B, C, D, or Fail grades.

2. **Subject-wise Score Distribution**  
   Shows the spread of marks in each subject.

3. **Top 10 Students**  
   Shows the students with the highest average scores.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Grade distribution
grade_counts = df["Grade"].value_counts().sort_index()
colors = ["#e74c3c", "#e67e22", "#f1c40f", "#2ecc71", "#3498db"]

axes[0].pie(
    grade_counts,
    labels=grade_counts.index,
    autopct="%1.1f%%",
    colors=colors[:len(grade_counts)],
    startangle=90
)
axes[0].set_title("Grade Distribution", fontsize=14, fontweight="bold")

# Subject-wise box plot
subject_cols = ["Math", "Science", "English", "History"]
df[subject_cols].boxplot(ax=axes[1])
axes[1].set_title("Subject Score Distribution", fontsize=14, fontweight="bold")
axes[1].set_ylabel("Marks")

# Top 10 students
top10 = df.nlargest(10, "Average")[["Name", "Average"]]

bars = axes[2].barh(top10["Name"], top10["Average"], color="#3498db")
axes[2].set_title("Top 10 Students by Average", fontsize=14, fontweight="bold")
axes[2].set_xlabel("Average Score")

for bar, val in zip(bars, top10["Average"]):
    axes[2].text(
        bar.get_width() + 0.5,
        bar.get_y() + bar.get_height() / 2,
        f"{val:.1f}",
        va="center",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

## 📋 Detailed Result Table

The table below shows the complete student result.

The students are sorted by average score in descending order.

Students who failed are highlighted in the Grade column.

```python
result_table = df[
    [
        "Student_ID",
        "Name",
        "Math",
        "Science",
        "English",
        "History",
        "Total",
        "Average",
        "Grade"
    ]
].sort_values("Average", ascending=False).reset_index(drop=True)

def highlight_fail(value):
    return "background-color: #ffcccc" if value == "Fail" else ""

styled_result = result_table.style.map(
    highlight_fail,
    subset=["Grade"]
).format({"Average": "{:.2f}"})

styled_result
```
